# 1. Introduction

Money laundering is a major financial crime that affects financial institutions, businesses, and economies worldwide. Financial institutions are required to monitor customer activity, identify unusual transaction patterns, investigate potential suspicious behaviour, and comply with Anti-Money Laundering (AML) requirements.

AML investigations can involve large amounts of transactional information and require analysts to combine data analysis with knowledge of AML concepts, regulatory requirements, and known money laundering typologies.

Recent developments in Deep Learning and Natural Language Processing have created new opportunities for applying Artificial Intelligence to financial crime detection and investigation. Transformer-based Language Models can process and generate natural language, while Retrieval-Augmented Generation (RAG) can provide models with relevant information retrieved from external knowledge sources.

This project explores how these technologies can be combined to create an AML Knowledge Assistant that supports analysts during the investigation process.

The proposed system is not intended to replace human investigators. Instead, it is designed as a decision-support tool that can help organize information, identify potential risk indicators, retrieve relevant AML knowledge, and generate structured investigation outputs.

# 2. Problem Statement

AML analysts often need to perform several tasks during an investigation:

- Review transaction activity.
- Identify unusual transaction patterns.
- Recognize potential AML red flags.
- Research AML concepts and regulatory requirements.
- Assess the potential risk associated with a case.
- Document investigation findings.
- Prepare structured reports.

Performing these activities manually can be time-consuming, particularly when the analyst needs to review large amounts of information or consult multiple sources.

The problem addressed by this project is therefore the development of an AI-based assistant that can combine transaction analysis with relevant AML knowledge and provide structured support during an investigation.

The main research problem can be expressed as follows:

> **Can a Transformer-based AI assistant combined with Retrieval-Augmented Generation (RAG) support AML analysts in analyzing transaction scenarios, identifying potential risk indicators, and retrieving relevant AML knowledge?**

The project will investigate the effectiveness of this approach using financial transaction data, AML-related documents, and predefined test cases.

The final system will be evaluated based on its ability to provide relevant information, identify potential AML indicators, and produce useful structured outputs for investigation support.

## 5.3 Retrieval-Augmented Generation

Retrieval-Augmented Generation combines information retrieval with text generation.

Instead of relying exclusively on information stored within a Language Model, a RAG system retrieves relevant information from an external knowledge base and provides this information to the model as additional context.

This approach is particularly relevant to AML because investigation processes often require analysts to consult external documentation and regulatory information.

The AML Knowledge Assistant therefore investigates whether adding a retrieval component can improve the relevance and reliability of generated AML-related responses.

# 6. Methodological Considerations and Evaluation Strategy

A key consideration in this project is the need for a realistic evaluation methodology for Anti-Money Laundering (AML) tasks.

AML transaction data are typically highly imbalanced and time-dependent. Therefore, model performance should not be evaluated using a simple random split or by artificially changing the class distribution without appropriate justification.

The following principles will be applied throughout the project.

## 6.1 Class Imbalance

The original class distribution will be preserved whenever possible.

Artificial oversampling will not be used simply to create a more balanced dataset. If sampling or resampling techniques are required for a specific experiment, the approach and its potential impact on evaluation results will be clearly documented.

Performance will be evaluated using minority-class metrics such as:

- Precision
- Recall
- F1-score
- False Positive Rate
- False Negative Rate
- Precision-Recall curves

Accuracy will not be used as the primary evaluation metric because it can be misleading when suspicious transactions represent a small proportion of the total transactions.

## 6.2 Time-Based Evaluation

Financial transactions are time-dependent. Therefore, the project will avoid relying exclusively on random train-test splitting.

Where timestamps are available, the data will be divided chronologically:

    Earlier Transactions
            ↓
         Training
            ↓
      Validation Period
            ↓
       Future Test Period

The final test set will represent a later period than the training data.

This approach is intended to provide a more realistic estimate of how the system may perform when applied to future transaction activity.

## 6.3 Account-Level Validation

Where the dataset structure allows it, evaluation will also consider the account level rather than evaluating only individual transactions.

This is important because transactions belonging to the same account can be highly related.

The project will investigate potential information leakage caused by the same accounts appearing across training and testing datasets.

Where appropriate, account-level splitting or additional account-level evaluation will be used.

## 6.4 Categorical Feature Encoding

Categorical variables will be encoded using methods appropriate for their statistical meaning and the selected model.

Categories will not be assigned arbitrary ordinal values when no natural ordering exists.

For example, countries such as Bulgaria, Germany, and France will not be represented as numerical values that could incorrectly imply an order between the countries.

The selected encoding method will be documented during the preprocessing stage.

## 6.5 Classification Threshold and Capacity Analysis

The project will not assume that a default probability threshold is automatically optimal.

Different classification thresholds will be evaluated to investigate the trade-off between:

- True Positives
- False Positives
- True Negatives
- False Negatives
- Precision
- Recall

The analysis will consider the operational impact of different thresholds and the potential investigation capacity of an AML team.

Where appropriate, the number of cases that would require investigation at different thresholds will also be examined.

## 6.6 Probability Calibration

Where the implemented model produces probability estimates, calibration will be investigated.

The purpose is to determine whether predicted risk probabilities correspond reasonably well to observed outcomes.

For example, cases assigned a predicted probability of approximately 0.80 should, ideally, contain a substantially higher proportion of positive cases than cases assigned a probability of approximately 0.20.

Calibration analysis will be performed where supported by the selected modelling approach and available test data.

## 6.7 False Positive and False Negative Analysis

The project will include a detailed analysis of incorrectly classified cases.

Examples will include:

- False Positives: normal activity incorrectly classified as suspicious.
- False Negatives: suspicious activity incorrectly classified as normal.

Representative cases will be examined to understand why the model produced an incorrect prediction.

This analysis is particularly important in AML because excessive false positives can increase investigation workload, while false negatives may result in potentially suspicious activity being missed.

## 6.8 Evaluation at Multiple Levels

The project will distinguish between transaction-level and account-level evaluation where possible.

Transaction-level evaluation will measure whether individual transactions are classified correctly.

Account-level evaluation will investigate whether the system can identify accounts associated with suspicious activity.

This provides a more complete view of the practical usefulness of the proposed approach.

## 6.9 Comparison of Evaluation Approaches

The project will compare different evaluation approaches where applicable, including:

- Random split versus time-based split
- Transaction-level versus account-level evaluation
- Different classification thresholds
- LLM without external retrieval versus LLM with RAG

The purpose is to determine whether the proposed AML Knowledge Assistant provides reliable and meaningful results under realistic evaluation conditions.

```python
import pandas as pd

data_path = found_files[0]

monthly_stats = []

for chunk in pd.read_csv(data_path, chunksize=500_000):
    
    chunk["Timestamp"] = pd.to_datetime(
        chunk["Timestamp"], 
        errors="coerce"
    )
    
    chunk["Month"] = chunk["Timestamp"].dt.to_period("M")
    
    monthly = chunk.groupby("Month").agg(
        Total_Transactions=("Is Laundering", "size"),
        Laundering_Transactions=("Is Laundering", "sum")
    ).reset_index()
    
    monthly_stats.append(monthly)

# Combine all chunks
monthly_stats = pd.concat(monthly_stats)

# Aggregate results
monthly_stats = (
    monthly_stats
    .groupby("Month")
    .sum()
    .reset_index()
)

monthly_stats["Laundering_Percentage"] = (
    monthly_stats["Laundering_Transactions"]
    / monthly_stats["Total_Transactions"]
    * 100
)

print("Monthly transaction distribution:")
display(monthly_stats)
```

```python
import pandas as pd
import os

# ============================================================
# CONTROLLED TIME-BASED SAMPLE
# ============================================================

data_path = r"C:\Users\Славена\Downloads\HI-Large_Trans.csv"

# Final sample sizes
TRAIN_SIZE = 100_000
VALIDATION_SIZE = 30_000
TEST_SIZE = 30_000

# Number of rows collected from each chunk
CHUNK_SAMPLE_SIZE = 5_000

# Storage for temporary samples
train_parts = []
validation_parts = []
test_parts = []

print("Starting controlled sampling...")
print("This may take a few minutes because the original file is very large.")
print()

for chunk_number, chunk in enumerate(
    pd.read_csv(data_path, chunksize=500_000),
    start=1
):

    # Convert timestamp
    chunk["Timestamp"] = pd.to_datetime(
        chunk["Timestamp"],
        errors="coerce"
    )

    # Remove invalid timestamps
    chunk = chunk.dropna(subset=["Timestamp"])

    # --------------------------------------------------------
    # TRAIN - AUGUST 2022
    # --------------------------------------------------------
    train_chunk = chunk[
        (chunk["Timestamp"] >= "2022-08-01") &
        (chunk["Timestamp"] < "2022-09-01")
    ]

    if len(train_chunk) > 0:
        n = min(CHUNK_SAMPLE_SIZE, len(train_chunk))

        train_parts.append(
            train_chunk.sample(
                n=n,
                random_state=42 + chunk_number
            )
        )

    # --------------------------------------------------------
    # VALIDATION - SEPTEMBER 2022
    # --------------------------------------------------------
    validation_chunk = chunk[
        (chunk["Timestamp"] >= "2022-09-01") &
        (chunk["Timestamp"] < "2022-10-01")
    ]

    if len(validation_chunk) > 0:
        n = min(CHUNK_SAMPLE_SIZE, len(validation_chunk))

        validation_parts.append(
            validation_chunk.sample(
                n=n,
                random_state=1000 + chunk_number
            )
        )

    # --------------------------------------------------------
    # TEST - OCTOBER 2022
    # --------------------------------------------------------
    test_chunk = chunk[
        (chunk["Timestamp"] >= "2022-10-01") &
        (chunk["Timestamp"] < "2022-11-01")
    ]

    if len(test_chunk) > 0:
        n = min(CHUNK_SAMPLE_SIZE, len(test_chunk))

        test_parts.append(
            test_chunk.sample(
                n=n,
                random_state=2000 + chunk_number
            )
        )

    # Progress message every 20 chunks
    if chunk_number % 20 == 0:
        print(f"Processed {chunk_number} chunks...")

# ============================================================
# COMBINE TEMPORARY SAMPLES
# ============================================================

train_candidates = pd.concat(
    train_parts,
    ignore_index=True
)

validation_candidates = pd.concat(
    validation_parts,
    ignore_index=True
)

test_candidates = pd.concat(
    test_parts,
    ignore_index=True
)

# ============================================================
# FINAL SAMPLE
# ============================================================

train_sample = train_candidates.sample(
    n=TRAIN_SIZE,
    random_state=42
).reset_index(drop=True)

validation_sample = validation_candidates.sample(
    n=VALIDATION_SIZE,
    random_state=42
).reset_index(drop=True)

test_sample = test_candidates.sample(
    n=TEST_SIZE,
    random_state=42
).reset_index(drop=True)

# ============================================================
# CREATE OUTPUT FOLDER
# ============================================================

output_folder = os.path.join(
    os.path.dirname(data_path),
    "AML_Knowledge_Assistant_Samples"
)

os.makedirs(output_folder, exist_ok=True)

# ============================================================
# SAVE FILES
# ============================================================

train_path = os.path.join(
    output_folder,
    "train_august.csv"
)

validation_path = os.path.join(
    output_folder,
    "validation_september.csv"
)

test_path = os.path.join(
    output_folder,
    "test_october.csv"
)

train_sample.to_csv(train_path, index=False)
validation_sample.to_csv(validation_path, index=False)
test_sample.to_csv(test_path, index=False)

# ============================================================
# FINAL REPORT
# ============================================================

print("\n" + "=" * 60)
print("SAMPLE CREATION COMPLETED")
print("=" * 60)

print("\nDataset sizes:")
print("Training:", f"{len(train_sample):,}")
print("Validation:", f"{len(validation_sample):,}")
print("Test:", f"{len(test_sample):,}")

print("\nLaundering distribution:")

print(
    "Training:",
    f"{train_sample['Is Laundering'].sum():,}",
    f"({train_sample['Is Laundering'].mean() * 100:.4f}%)"
)

print(
    "Validation:",
    f"{validation_sample['Is Laundering'].sum():,}",
    f"({validation_sample['Is Laundering'].mean() * 100:.4f}%)"
)

print(
    "Test:",
    f"{test_sample['Is Laundering'].sum():,}",
    f"({test_sample['Is Laundering'].mean() * 100:.4f}%)"
)

print("\nSaved files:")
print(train_path)
print(validation_path)
print(test_path)
```

```python

import pandas as pd
import numpy as np
import os

# ============================================================
# FINAL REPRESENTATIVE TIME-BASED SAMPLE
# ============================================================

data_path = r"C:\Users\Славена\Downloads\HI-Large_Trans.csv"

# Final sample sizes
TRAIN_SIZE = 500_000
VALIDATION_SIZE = 250_000
TEST_SIZE = 250_000

# Original monthly transaction counts
AUG_TOTAL = 59_461_836
SEP_TOTAL = 56_137_121
OCT_TOTAL = 54_821_376

# Small safety buffer to make sure enough candidates are collected
BUFFER = 1.05

# Sampling probabilities
AUG_RATE = (TRAIN_SIZE * BUFFER) / AUG_TOTAL
SEP_RATE = (VALIDATION_SIZE * BUFFER) / SEP_TOTAL
OCT_RATE = (TEST_SIZE * BUFFER) / OCT_TOTAL

print("Sampling rates:")
print(f"August:    {AUG_RATE:.6%}")
print(f"September: {SEP_RATE:.6%}")
print(f"October:   {OCT_RATE:.6%}")

train_parts = []
validation_parts = []
test_parts = []

rng = np.random.default_rng(42)

# ============================================================
# READ LARGE FILE IN CHUNKS
# ============================================================

for chunk_number, chunk in enumerate(
    pd.read_csv(data_path, chunksize=500_000),
    start=1
):

    chunk["Timestamp"] = pd.to_datetime(
        chunk["Timestamp"],
        errors="coerce"
    )

    chunk = chunk.dropna(subset=["Timestamp"])

    # --------------------------------------------------------
    # AUGUST - TRAIN
    # --------------------------------------------------------

    august = chunk[
        (chunk["Timestamp"] >= "2022-08-01") &
        (chunk["Timestamp"] < "2022-09-01")
    ]

    if len(august) > 0:
        mask = rng.random(len(august)) < AUG_RATE
        selected = august.loc[mask]

        if len(selected) > 0:
            train_parts.append(selected)

    # --------------------------------------------------------
    # SEPTEMBER - VALIDATION
    # --------------------------------------------------------

    september = chunk[
        (chunk["Timestamp"] >= "2022-09-01") &
        (chunk["Timestamp"] < "2022-10-01")
    ]

    if len(september) > 0:
        mask = rng.random(len(september)) < SEP_RATE
        selected = september.loc[mask]

        if len(selected) > 0:
            validation_parts.append(selected)

    # --------------------------------------------------------
    # OCTOBER - TEST
    # --------------------------------------------------------

    october = chunk[
        (chunk["Timestamp"] >= "2022-10-01") &
        (chunk["Timestamp"] < "2022-11-01")
    ]

    if len(october) > 0:
        mask = rng.random(len(october)) < OCT_RATE
        selected = october.loc[mask]

        if len(selected) > 0:
            test_parts.append(selected)

    # Progress
    if chunk_number % 20 == 0:
        print(f"Processed {chunk_number} chunks...")


# ============================================================
# COMBINE CANDIDATES
# ============================================================

train_candidates = pd.concat(
    train_parts,
    ignore_index=True
)

validation_candidates = pd.concat(
    validation_parts,
    ignore_index=True
)

test_candidates = pd.concat(
    test_parts,
    ignore_index=True
)

print("\nCandidates collected:")
print("Train:", f"{len(train_candidates):,}")
print("Validation:", f"{len(validation_candidates):,}")
print("Test:", f"{len(test_candidates):,}")


# ============================================================
# CHECK THAT ENOUGH CANDIDATES EXIST
# ============================================================

if len(train_candidates) < TRAIN_SIZE:
    raise ValueError(
        f"Not enough training candidates: "
        f"{len(train_candidates):,} available, "
        f"{TRAIN_SIZE:,} required."
    )

if len(validation_candidates) < VALIDATION_SIZE:
    raise ValueError(
        f"Not enough validation candidates: "
        f"{len(validation_candidates):,} available, "
        f"{VALIDATION_SIZE:,} required."
    )

if len(test_candidates) < TEST_SIZE:
    raise ValueError(
        f"Not enough test candidates: "
        f"{len(test_candidates):,} available, "
        f"{TEST_SIZE:,} required."
    )


# ============================================================
# FINAL EXACT SAMPLE SIZE
# ============================================================

train_sample = train_candidates.sample(
    n=TRAIN_SIZE,
    random_state=42,
    replace=False
).reset_index(drop=True)

validation_sample = validation_candidates.sample(
    n=VALIDATION_SIZE,
    random_state=42,
    replace=False
).reset_index(drop=True)

test_sample = test_candidates.sample(
    n=TEST_SIZE,
    random_state=42,
    replace=False
).reset_index(drop=True)


# ============================================================
# SAVE FINAL DATASETS
# ============================================================

output_folder = (
    r"C:\Users\Славена\Downloads"
    r"\AML_Knowledge_Assistant_Samples_Final"
)

os.makedirs(output_folder, exist_ok=True)

train_path = os.path.join(
    output_folder,
    "train_august.csv"
)

validation_path = os.path.join(
    output_folder,
    "validation_september.csv"
)

test_path = os.path.join(
    output_folder,
    "test_october.csv"
)

train_sample.to_csv(
    train_path,
    index=False
)

validation_sample.to_csv(
    validation_path,
    index=False
)

test_sample.to_csv(
    test_path,
    index=False
)


# ============================================================
# FINAL DISTRIBUTION CHECK
# ============================================================

print("\n" + "=" * 60)
print("FINAL SAMPLE CREATION COMPLETED")
print("=" * 60)

for name, df in [
    ("TRAIN - AUGUST", train_sample),
    ("VALIDATION - SEPTEMBER", validation_sample),
    ("TEST - OCTOBER", test_sample)
]:

    laundering = df["Is Laundering"].sum()
    percentage = df["Is Laundering"].mean() * 100

    print(f"\n{name}")
    print("Rows:", f"{len(df):,}")
    print("Laundering:", f"{laundering:,}")
    print("Laundering %:", f"{percentage:.4f}%")


# ============================================================
# SAVED FILES
# ============================================================

print("\nSaved files:")
print(train_path)
print(validation_path)
print(test_path)
```

```python
# ============================================================
# FIX - CREATE RELATIONSHIP FEATURES IN ALL DATASETS
# ============================================================

for df in [train_df, validation_df, test_df]:

    # Same bank
    df["Same Bank"] = (
        df["From Bank"] ==
        df["To Bank"]
    ).astype(int)

    # Cross bank
    df["Cross Bank"] = (
        df["From Bank"] !=
        df["To Bank"]
    ).astype(int)

    # Same currency
    df["Same Currency"] = (
        df["Receiving Currency"] ==
        df["Payment Currency"]
    ).astype(int)

print("=" * 60)
print("RELATIONSHIP FEATURES FIXED")
print("=" * 60)

for df_name, df in [
    ("TRAIN", train_df),
    ("VALIDATION", validation_df),
    ("TEST", test_df)
]:
    print(
        f"{df_name}: "
        f"Same Bank={('Same Bank' in df.columns)}, "
        f"Cross Bank={('Cross Bank' in df.columns)}, "
        f"Same Currency={('Same Currency' in df.columns)}"
    )
```

```python
# ============================================================
# PREPROCESSING - ONE-HOT ENCODING + SCALING
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical features
numeric_features = [
    "From Bank",
    "To Bank",
    "Amount Received",
    "Amount Paid",
    "Hour",
    "Day",
    "Day of Week",
    "Month",
    "Is Weekend",
    "Same Bank",
    "Cross Bank"
]

# Categorical features
categorical_features = [
    "Receiving Currency",
    "Payment Currency",
    "Payment Format"
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

# Separate X and y
X_train = train_df[numeric_features + categorical_features]
y_train = train_df[TARGET]

X_validation = validation_df[numeric_features + categorical_features]
y_validation = validation_df[TARGET]

X_test = test_df[numeric_features + categorical_features]
y_test = test_df[TARGET]

print("=" * 60)
print("PREPROCESSING OBJECT CREATED")
print("=" * 60)

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Total input features:", len(numeric_features) + len(categorical_features))

print("\nTrain:", X_train.shape)
print("Validation:", X_validation.shape)
print("Test:", X_test.shape)
```

```python
# ============================================================
# PREPROCESSING - FIT ON TRAIN ONLY
# ============================================================

print("=" * 60)
print("FITTING PREPROCESSOR ON TRAINING DATA")
print("=" * 60)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

print("Train preprocessing completed.")

# Transform validation and test using the TRAIN-fitted preprocessor
X_validation_processed = preprocessor.transform(X_validation)
X_test_processed = preprocessor.transform(X_test)

print("Validation preprocessing completed.")
print("Test preprocessing completed.")

print("\nProcessed shapes:")
print("Train:", X_train_processed.shape)
print("Validation:", X_validation_processed.shape)
print("Test:", X_test_processed.shape)

print("\nTarget shapes:")
print("Train:", y_train.shape)
print("Validation:", y_validation.shape)
print("Test:", y_test.shape)

print("\nPreprocessing completed successfully.")
```

```python
# ============================================================
# MODEL 1 - LOGISTIC REGRESSION BASELINE
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

print("=" * 60)
print("LOGISTIC REGRESSION - TRAINING")
print("=" * 60)

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

logistic_model.fit(
    X_train_processed,
    y_train
)

print("Logistic Regression trained successfully.")

# Validation predictions
y_val_pred = logistic_model.predict(X_validation_processed)
y_val_prob = logistic_model.predict_proba(
    X_validation_processed
)[:, 1]

# Evaluation
logistic_precision = precision_score(
    y_validation,
    y_val_pred,
    zero_division=0
)

logistic_recall = recall_score(
    y_validation,
    y_val_pred,
    zero_division=0
)

logistic_f1 = f1_score(
    y_validation,
    y_val_pred,
    zero_division=0
)

logistic_roc_auc = roc_auc_score(
    y_validation,
    y_val_prob
)

logistic_pr_auc = average_precision_score(
    y_validation,
    y_val_prob
)

print("\nVALIDATION RESULTS")
print("-" * 40)
print(f"Precision: {logistic_precision:.4f}")
print(f"Recall:    {logistic_recall:.4f}")
print(f"F1-score:  {logistic_f1:.4f}")
print(f"ROC-AUC:   {logistic_roc_auc:.4f}")
print(f"PR-AUC:    {logistic_pr_auc:.4f}")
```

```python
# ============================================================
# MODEL 2 - DECISION TREE
# ============================================================

from sklearn.tree import DecisionTreeClassifier

print("=" * 60)
print("DECISION TREE - TRAINING")
print("=" * 60)

decision_tree_model = DecisionTreeClassifier(
    max_depth=8,
    min_samples_leaf=50,
    class_weight="balanced",
    random_state=42
)

decision_tree_model.fit(
    X_train_processed,
    y_train
)

print("Decision Tree trained successfully.")

# Validation predictions
y_val_pred_tree = decision_tree_model.predict(
    X_validation_processed
)

y_val_prob_tree = decision_tree_model.predict_proba(
    X_validation_processed
)[:, 1]

# Evaluation
tree_precision = precision_score(
    y_validation,
    y_val_pred_tree,
    zero_division=0
)

tree_recall = recall_score(
    y_validation,
    y_val_pred_tree,
    zero_division=0
)

tree_f1 = f1_score(
    y_validation,
    y_val_pred_tree,
    zero_division=0
)

tree_roc_auc = roc_auc_score(
    y_validation,
    y_val_prob_tree
)

tree_pr_auc = average_precision_score(
    y_validation,
    y_val_prob_tree
)

print("\nVALIDATION RESULTS")
print("-" * 40)
print(f"Precision: {tree_precision:.4f}")
print(f"Recall:    {tree_recall:.4f}")
print(f"F1-score:  {tree_f1:.4f}")
print(f"ROC-AUC:   {tree_roc_auc:.4f}")
print(f"PR-AUC:    {tree_pr_auc:.4f}")
```

```python
# ============================================================
# MODEL 3 - RANDOM FOREST
# ============================================================

from sklearn.ensemble import RandomForestClassifier

print("=" * 60)
print("RANDOM FOREST - TRAINING")
print("=" * 60)

random_forest_model = RandomForestClassifier(
    n_estimators=150,
    max_depth=12,
    min_samples_leaf=20,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(
    X_train_processed,
    y_train
)

print("Random Forest trained successfully.")

# Validation predictions
y_val_pred_rf = random_forest_model.predict(
    X_validation_processed
)

y_val_prob_rf = random_forest_model.predict_proba(
    X_validation_processed
)[:, 1]

# Evaluation
rf_precision = precision_score(
    y_validation,
    y_val_pred_rf,
    zero_division=0
)

rf_recall = recall_score(
    y_validation,
    y_val_pred_rf,
    zero_division=0
)

rf_f1 = f1_score(
    y_validation,
    y_val_pred_rf,
    zero_division=0
)

rf_roc_auc = roc_auc_score(
    y_validation,
    y_val_prob_rf
)

rf_pr_auc = average_precision_score(
    y_validation,
    y_val_prob_rf
)

print("\nVALIDATION RESULTS")
print("-" * 40)
print(f"Precision: {rf_precision:.4f}")
print(f"Recall:    {rf_recall:.4f}")
print(f"F1-score:  {rf_f1:.4f}")
print(f"ROC-AUC:   {rf_roc_auc:.4f}")
print(f"PR-AUC:    {rf_pr_auc:.4f}")
```

```python
# ============================================================
# AML CAPACITY / TOP-K THRESHOLD ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

# Create evaluation dataframe
capacity_df = pd.DataFrame({
    "Actual": y_validation.to_numpy(),
    "Risk_Score": y_val_prob_rf
})

# Sort transactions from highest to lowest predicted risk
capacity_df = capacity_df.sort_values(
    "Risk_Score",
    ascending=False
).reset_index(drop=True)

# Capacity levels
capacity_levels = [
    100,
    500,
    1000,
    2500,
    5000,
    10000
]

capacity_results = []

total_laundering = capacity_df["Actual"].sum()

for capacity in capacity_levels:

    top_k = capacity_df.head(capacity)

    detected = top_k["Actual"].sum()

    precision = detected / capacity
    recall = detected / total_laundering

    threshold = top_k["Risk_Score"].min()

    capacity_results.append({
        "Review Capacity": capacity,
        "Detected Laundering": int(detected),
        "Precision": precision,
        "Recall": recall,
        "Risk Threshold": threshold
    })

capacity_results = pd.DataFrame(capacity_results)

print("=" * 80)
print("RANDOM FOREST - AML CAPACITY / TOP-K ANALYSIS")
print("=" * 80)

display(
    capacity_results.style.format({
        "Precision": "{:.2%}",
        "Recall": "{:.2%}",
        "Risk Threshold": "{:.4f}"
    })
)
```

```python
# ============================================================
# THRESHOLD ANALYSIS - RANDOM FOREST
# ============================================================

thresholds_to_test = [
    0.50,
    0.40,
    0.30,
    0.20,
    0.10,
    0.05,
    0.01
]

threshold_results = []

for threshold in thresholds_to_test:

    predictions = (
        y_val_prob_rf >= threshold
    ).astype(int)

    precision = precision_score(
        y_validation,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_validation,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_validation,
        predictions,
        zero_division=0
    )

    flagged = predictions.sum()

    threshold_results.append({
        "Threshold": threshold,
        "Flagged Transactions": int(flagged),
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

threshold_results = pd.DataFrame(threshold_results)

print("=" * 80)
print("RANDOM FOREST - THRESHOLD ANALYSIS")
print("=" * 80)

display(
    threshold_results.style.format({
        "Threshold": "{:.2f}",
        "Precision": "{:.2%}",
        "Recall": "{:.2%}",
        "F1-Score": "{:.4f}"
    })
)
```

```python
# ============================================================
# RAG - RETRIEVAL TEST
# ============================================================

def retrieve_documents(query, top_k=3):

    # Convert query into embedding
    query_embedding = embedding_model.encode(
        [query]
    ).astype("float32")

    # Search FAISS
    distances, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    results = []

    for distance, index in zip(distances[0], indices[0]):

        results.append({
            "Document_Index": int(index),
            "Distance": float(distance),
            "Document": documents[index]
        })

    return results


# ============================================================
# TEST QUERY
# ============================================================

query = "What are the Belgian AML requirements for atypical transactions?"

results = retrieve_documents(
    query,
    top_k=3
)

print("=" * 60)
print("RAG RETRIEVAL TEST")
print("=" * 60)

print("\nQuery:")
print(query)

for i, result in enumerate(results, 1):

    print("\n" + "-" * 60)
    print(f"RESULT {i}")
    print("-" * 60)

    print("Document index:", result["Document_Index"])
    print("Distance:", round(result["Distance"], 4))

    print("\nRetrieved document:")
    print(result["Document"])
```

```python
# ============================================================
# AML KNOWLEDGE ASSISTANT - RETRIEVAL FUNCTION
# ============================================================

def aml_retrieve(query, top_k=3):

    query_embedding = embedding_model.encode(
        [query]
    ).astype("float32")

    distances, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    retrieved = []

    for rank, (distance, index) in enumerate(
        zip(distances[0], indices[0]), start=1
    ):

        retrieved.append({
            "Rank": rank,
            "Document_Index": int(index),
            "Distance": float(distance),
            "Topic": belgian_aml_kb.iloc[index]["Topic"],
            "Rule": belgian_aml_kb.iloc[index]["Rule"],
            "Legal_Reference": belgian_aml_kb.iloc[index]["Legal_Reference"],
            "Source": belgian_aml_kb.iloc[index]["Source"],
            "Practical_Interpretation": belgian_aml_kb.iloc[index]["Practical_Interpretation"]
        })

    return retrieved


# ============================================================
# TEST AML KNOWLEDGE ASSISTANT
# ============================================================

question = "What should an AML analyst do when a transaction is considered atypical?"

retrieved = aml_retrieve(
    question,
    top_k=3
)

print("=" * 70)
print("AML KNOWLEDGE ASSISTANT - RETRIEVAL")
print("=" * 70)

print("\nQuestion:")
print(question)

for item in retrieved:

    print("\n" + "-" * 70)
    print(f"RESULT {item['Rank']}")
    print("-" * 70)

    print("Topic:", item["Topic"])
    print("Legal Reference:", item["Legal_Reference"])
    print("Source:", item["Source"])

    print("\nRule:")
    print(item["Rule"])

    print("\nPractical Interpretation:")
    print(item["Practical_Interpretation"])
```

```python
# ============================================================
# AML KNOWLEDGE ASSISTANT - TRANSACTION / CASE ANALYZER
# ============================================================

def analyze_aml_case(
    amount_received,
    number_of_senders,
    rapid_outgoing_transfer,
    profile_mismatch
):

    red_flags = []

    # --------------------------------------------------------
    # 1. RED FLAGS
    # --------------------------------------------------------

    if amount_received >= 50000:
        red_flags.append(
            f"High-value transaction: €{amount_received:,.2f}"
        )

    if number_of_senders >= 3:
        red_flags.append(
            f"Multiple unrelated senders: {number_of_senders}"
        )

    if rapid_outgoing_transfer:
        red_flags.append(
            "Rapid movement of funds after receipt"
        )

    if profile_mismatch:
        red_flags.append(
            "Transaction activity inconsistent with customer profile"
        )

    # --------------------------------------------------------
    # 2. RISK LEVEL
    # --------------------------------------------------------

    risk_score = len(red_flags)

    if risk_score >= 3:
        risk_level = "HIGH"
    elif risk_score == 2:
        risk_level = "MEDIUM"
    else:
        risk_level = "LOW"

    # --------------------------------------------------------
    # 3. RAG RETRIEVAL
    # --------------------------------------------------------

    case_question = """
    What should an AML analyst do when transactions are unusually
    large, involve multiple senders, move rapidly, or are inconsistent
    with the customer's expected activity?
    """

    retrieved = aml_retrieve(
        case_question,
        top_k=3
    )

    # --------------------------------------------------------
    # 4. DISPLAY CASE ANALYSIS
    # --------------------------------------------------------

    print("=" * 70)
    print("AML TRANSACTION / CASE ANALYZER")
    print("=" * 70)

    print("\nCASE PROFILE")
    print("-" * 70)
    print(f"Amount received: €{amount_received:,.2f}")
    print(f"Number of senders: {number_of_senders}")
    print(f"Rapid outgoing transfer: {rapid_outgoing_transfer}")
    print(f"Profile mismatch: {profile_mismatch}")

    print("\nRISK ASSESSMENT")
    print("-" * 70)
    print(f"Risk Score: {risk_score}")
    print(f"Risk Level: {risk_level}")

    print("\nRED FLAGS")
    print("-" * 70)

    if red_flags:
        for i, flag in enumerate(red_flags, 1):
            print(f"{i}. {flag}")
    else:
        print("No significant red flags identified.")

    print("\nAPPLICABLE BELGIAN AML KNOWLEDGE")
    print("-" * 70)

    for item in retrieved:

        print(f"\n• {item['Topic']}")
        print(f"  Rule: {item['Rule']}")
        print(f"  Legal Reference: {item['Legal_Reference']}")
        print(f"  Source: {item['Source']}")

    print("\nRECOMMENDED ACTIONS")
    print("-" * 70)

    if risk_level == "HIGH":
        print("1. Perform enhanced case analysis.")
        print("2. Review the customer's expected activity and profile.")
        print("3. Review the source and movement of funds.")
        print("4. Escalate the atypical activity for AMLCO analysis.")
    elif risk_level == "MEDIUM":
        print("1. Perform additional transaction analysis.")
        print("2. Compare activity with the customer's expected profile.")
        print("3. Consider escalation if concerns remain.")
    else:
        print("1. Continue normal transaction monitoring.")
        print("2. Document the assessment where appropriate.")

    print("\nIMPORTANT")
    print("-" * 70)
    print(
        "A red flag or atypical transaction does not by itself prove "
        "money laundering or terrorist financing."
    )

    print("\n" + "=" * 70)


# ============================================================
# TEST CASE
# ============================================================

analyze_aml_case(
    amount_received=85000,
    number_of_senders=5,
    rapid_outgoing_transfer=True,
    profile_mismatch=True
)
```

```python
# ============================================================
# LLM vs LLM + RAG - COMPARISON FRAMEWORK
# ============================================================

print("=" * 70)
print("LLM vs LLM + RAG COMPARISON")
print("=" * 70)

# ------------------------------------------------------------
# TEST QUESTIONS
# ------------------------------------------------------------

evaluation_questions = [
    {
        "question": "What should an AML analyst do when a transaction is considered atypical?",
        "expected_topics": [
            "Atypical Transactions",
            "Atypical Transaction Indicators",
            "AMLCO Escalation"
        ]
    },
    {
        "question": "What Belgian AML requirements apply to transaction monitoring?",
        "expected_topics": [
            "Ongoing Transaction Monitoring",
            "Risk-Based Approach"
        ]
    },
    {
        "question": "When should an atypical transaction be escalated?",
        "expected_topics": [
            "AMLCO Escalation",
            "Atypical Transactions"
        ]
    }
]


# ------------------------------------------------------------
# BASELINE RESPONSE
# ------------------------------------------------------------

def baseline_response(question):

    return (
        "The transaction should be reviewed for unusual activity, "
        "customer profile inconsistencies and potential AML risks. "
        "Further investigation may be appropriate depending on the "
        "circumstances."
    )


# ------------------------------------------------------------
# RAG RESPONSE
# ------------------------------------------------------------

def rag_response(question, top_k=3):

    retrieved = aml_retrieve(
        question,
        top_k=top_k
    )

    response = []

    for item in retrieved:

        response.append({
            "Topic": item["Topic"],
            "Rule": item["Rule"],
            "Legal Reference": item["Legal_Reference"],
            "Source": item["Source"],
            "Practical Interpretation": item["Practical_Interpretation"]
        })

    return response


# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

comparison_results = []

for test in evaluation_questions:

    question = test["question"]
    expected_topics = test["expected_topics"]

    retrieved = rag_response(question)

    retrieved_topics = [
        item["Topic"]
        for item in retrieved
    ]

    matched_topics = [
        topic
        for topic in expected_topics
        if topic in retrieved_topics
    ]

    baseline = baseline_response(question)

    # Simple transparent evaluation
    baseline_has_legal_reference = False

    rag_has_legal_reference = any(
        item["Legal Reference"]
        for item in retrieved
    )

    rag_topic_coverage = (
        len(matched_topics) / len(expected_topics)
    )

    comparison_results.append({
        "Question": question,
        "Baseline Legal Grounding": int(
            baseline_has_legal_reference
        ),
        "RAG Legal Grounding": int(
            rag_has_legal_reference
        ),
        "RAG Topic Coverage": rag_topic_coverage,
        "Retrieved Topics": ", ".join(retrieved_topics)
    })


comparison_df = pd.DataFrame(comparison_results)


# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPARISON RESULTS")
print("=" * 70)

display(comparison_df)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

print(
    "Baseline responses contain no explicit Belgian legal references."
)

print(
    "RAG responses retrieve Belgian AML topics, legal references "
    "and practical interpretations from the Knowledge Base."
)

print(
    f"\nAverage RAG topic coverage: "
    f"{comparison_df['RAG Topic Coverage'].mean():.2%}"
)
```

This suggests that the optimal threshold should not be selected solely according to a statistical metric. It should also consider investigation capacity, risk appetite and the requirements of a risk-based AML monitoring framework.

The feature importance analysis showed that payment format and transaction amounts were among the strongest predictive features.

## 4. Contribution of RAG

The RAG component provides a different type of value from the machine learning model.

The machine learning model focuses on identifying potentially suspicious transaction patterns, while RAG provides contextual AML knowledge that can support the interpretation of those patterns.

The RAG evaluation achieved 100% legal grounding and 100% topic coverage across the three predefined test questions. The system retrieved relevant Belgian AML topics, legal references and practical interpretations.

This demonstrates the potential benefit of combining predictive analytics with a dedicated AML knowledge base.

## 5. Belgian AML Context

**Transaction → Red Flags → Risk Level → RAG Retrieval → Belgian AML Knowledge → Recommended Actions → SAR Draft**

# Limitations

Although the proposed AML system demonstrates promising results, several limitations should be considered.

## 1. Dataset Limitations

The machine learning experiments were conducted using synthetic AML transaction datasets and sampled data. Synthetic data may not fully represent the complexity, variability and behavioural patterns observed in real-world financial transactions.

The results should therefore not be interpreted as direct evidence of performance in a production banking environment.

## 2. Class Imbalance

AML datasets are highly imbalanced, with legitimate transactions substantially outnumbering laundering transactions.

This imbalance contributed to the relatively low Precision and F1-scores observed across the evaluated models. Although ROC-AUC values were relatively strong, ROC-AUC alone does not provide a complete picture of performance in a highly imbalanced AML classification problem.

For this reason, Precision, Recall, F1-score, PR-AUC and operational review capacity were considered together.

## 3. False Positives

The Random Forest model generated a significant number of false positives.

In a real AML environment, false positives can increase investigation workload and operational costs. The model therefore should be used to prioritise cases rather than automatically classify customers or transactions as involved in financial crime.

## 4. Limited Knowledge Base

The Belgian AML Knowledge Base used in the RAG component contains a limited number of documents and topics.

Although the retrieval system successfully returned relevant information during the evaluation, a production system would require a much broader and continuously maintained knowledge base covering applicable legislation, regulatory guidance, institutional procedures and relevant updates.

## 5. RAG Evaluation Size

The RAG evaluation was performed using three predefined questions.

The system achieved 100% legal grounding and 100% topic coverage for these questions. However, this evaluation set is small and cannot establish general performance across all possible AML questions.

A larger evaluation dataset with independently validated questions and answers would provide a more robust assessment.

## 6. No Full Generative LLM Benchmark

The project evaluates a baseline response against RAG-based AML knowledge retrieval. The baseline is a generic predefined response rather than a separate production-scale generative LLM.

Therefore, the comparison should not be interpreted as a complete benchmark between two independent LLM systems.

A future version could compare the same generative model with and without RAG using a larger set of AML questions and human-validated evaluation criteria.

## 7. Human Oversight

The system is designed as an AML decision-support prototype.

Risk levels, red flags, retrieved legal information and generated SAR drafts should be reviewed by qualified AML professionals. The system should not independently determine that money laundering or terrorist financing has occurred.

The SAR component produces a draft for analyst or AMLCO review rather than an automatically submitted report.

## 8. Legal and Regulatory Updates

AML legislation, regulatory guidance and supervisory expectations may change over time.

The knowledge base would therefore require regular review and updating to ensure that retrieved information remains appropriate for the intended Belgian AML context.

## 9. Production Deployment

The prototype has not been evaluated in a live banking infrastructure.

A production implementation would require additional considerations including data security, privacy, access controls, auditability, model monitoring, explainability, governance and integration with existing AML transaction monitoring systems.

## Overall Limitation

The project should therefore be interpreted as a research and proof-of-concept system demonstrating how machine learning and retrieval-augmented AML knowledge can be combined.

Further validation using real-world data, a larger legal knowledge base, broader RAG evaluation and human expert assessment would be required before considering production deployment.

Machine learning can support the identification and prioritisation of potentially suspicious transactions, while RAG can provide contextual Belgian AML information to support analyst decision-making.

The proposed solution should be considered a research prototype and decision-support tool rather than an autonomous AML decision system. Human oversight remains essential, particularly when interpreting alerts, assessing suspicious activity and preparing regulatory reports.

Future development could include a larger and continuously updated Belgian AML knowledge base, real-world transaction data, a larger RAG evaluation dataset, human expert validation and a genuine generative LLM comparison with and without RAG. These improvements could further evaluate the system's effectiveness and suitability for more realistic AML investigation environments.
